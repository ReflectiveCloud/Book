Reflective's unified Python interface for accessing SAI (Stratospheric Aerosol Injection) climate model data across cloud providers (S3, GCS, Azure, Cloudflare R2) for use on the [Reflective Cloud Hub](https://reflective.2i2c.cloud).
Note: At this time, most non ESGF and ESM sources will not be accessible outside of the Reflective Cloud Hub due to technical limitations. We are working on storing the data in new locations and will update when it's ready.

This is an [Intake](https://github.com/intake/intake)-like interface for browsing, searching, and loading SRM related datasets in a unified manner. All available datasets [can be seen here](https://docs.google.com/spreadsheets/d/1cjgJQSrDV_IQVN68HoTQpy_xGQQhPy4wz1N0u8E2Pe4/edit?usp=sharing).

```python
from reflective_data_catalog import ReflectiveCatalog
rdc = ReflectiveCatalog()
```

Note that once this is fully incorporated into the Cloud Hub, we will not have to do the above steps.

## Using the Reflective Data Catalog

See what sources are available with:

```python
rdc.list_sources()
```

Search for data with

```python
rdc.search(term='ssp245')
rdc.search(variable='tas')
rdc.search(tag='SAI')
rdc.search(term='ssp245', variable='tas')
```

Learn more about a specific data source

```python
rdc.cesm2_waccm_g6_1p5k_hilla().discover()
```

See what variables exist for specific ensemble and table combinations

```python
rdc.cesm2_waccm_g6_1p5k_hilla().list_variables(ensemble='r2', table='ADAY') 
```

Loading data

**Note:** This can make your instance OOM if you have selected one of the smaller ones or try to read in daily data.

```python
ds = rdc.cesm2_waccm_g6_1p5k_hilla().to_dask() # Lazy loading
```

```python
ds2 = rdc.cesm2_waccm_g6_1p5k_hilla().read() # to memory
```

## Using with ESGF sources

Search for ESGF data only

```python
rdc.esgf.search(project='CMIP6',
                experiment_id='G6solar',
                source_id='UKESM1-0-LL',
                variable_id='tas'
               )
```

### Load GEOMIP G6 Sulfar data

```python
geomip_g6Sulfur_ds = rdc.esgf.geomip.g6sulfur()
geomip_g6Sulfur_ds
```

### Load a specific SSP

```python
ssp245_ds = rdc.esgf.ssp.ssp245()
```

It's (`rdc.esgf`) [intake-esgf](https://intake-esgf.readthedocs.io/en/latest/) under the hood and all functionality can still be used.

## Using with ESM sources
Search and load in one step

```python
ds = rdc.esm.load(
    experiment_id=['G6sulfur', 'ssp245', 'ssp585'],
    variable_id='tas',
    table_id='Amon',
    require_all_on=['source_id', 'institution_id'],
)
```

or use the GeoMIP convience helper

```python
ds = rdc.geomip_cloud.load_ensemble(
    experiments=['G6sulfur', 'ssp245', 'ssp585'],
    variable='tas',
)
```

Quickly load a single experiment

```python
ds_dict = rdc.geomip_cloud.g6sulfur(variable='tas')
```

Explore what is available 

```python
rdc.geomip_cloud.list_models()
rdc.geomip_cloud.list_variables(experiment_id='G6sulfur')
rdc.geomip_cloud.summary()
```

direct search then load

```python
subset = rdc.esm.search(
    experiment_id='G6sulfur',
    variable_id=['tas'],
    table_id='Amon',
)
datasets = subset.to_dataset_dict()
```

Just like with ESGF, this (`rdc.esm`) is [intake-esm](https://intake-esm.readthedocs.io/en/stable/) under the hood and all of it's functionality is available to use.
